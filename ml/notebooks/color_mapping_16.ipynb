## 4. Konversi warna

Notebook memakai konversi sRGB ke HSV dan CIELAB D65. Nilai Lab dipakai untuk jarak warna di aplikasi.


```python
def hex_to_rgb(hex_code):
    h = hex_code.lstrip("#")
    return tuple(int(h[i:i+2], 16) for i in (0, 2, 4))


def rgb_to_hsv(r, g, b):
    r, g, b = r / 255.0, g / 255.0, b / 255.0
    mx, mn = max(r, g, b), min(r, g, b)
    d = mx - mn
    if d == 0:
        h = 0.0
    elif mx == r:
        h = (60 * ((g - b) / d)) % 360
    elif mx == g:
        h = 60 * ((b - r) / d) + 120
    else:
        h = 60 * ((r - g) / d) + 240
    s = 0.0 if mx == 0 else d / mx
    return h, s, mx


def rgb_to_lab(rgb):
    # sRGB 0..255 -> linear sRGB -> XYZ D65 -> Lab
    c = np.array(rgb, dtype=np.float64) / 255.0
    c = np.where(c <= 0.04045, c / 12.92, ((c + 0.055) / 1.055) ** 2.4)
    x, y, z = (
        0.4124564 * c[0] + 0.3575761 * c[1] + 0.1804375 * c[2],
        0.2126729 * c[0] + 0.7151522 * c[1] + 0.0721750 * c[2],
        0.0193339 * c[0] + 0.1191920 * c[1] + 0.9503041 * c[2],
    )
    x, y, z = x / 0.95047, y / 1.0, z / 1.08883
    delta = 6 / 29

    def f(t):
        return t ** (1 / 3) if t > delta ** 3 else t / (3 * delta ** 2) + 4 / 29

    fx, fy, fz = f(x), f(y), f(z)
    return 116 * fy - 16, 500 * (fx - fy), 200 * (fy - fz)


def color_rows():
    aliases = {}
    for base, family in BASE_TO_FAMILY.items():
        aliases.setdefault(family, []).append(base)

    rows = []
    for key, hex_code, tone, neutral, legacy in COLOR_FAMILIES:
        r, g, b = hex_to_rgb(hex_code)
        h, s, v = rgb_to_hsv(r, g, b)
        lab_l, lab_a, lab_b = rgb_to_lab((r, g, b))
        rows.append({
            "color_family": key,
            "hex": hex_code,
            "r": r,
            "g": g,
            "b": b,
            "h": round(h, 2),
            "s": round(s, 4),
            "v": round(v, 4),
            "lab_l": round(lab_l, 2),
            "lab_a": round(lab_a, 2),
            "lab_b": round(lab_b, 2),
            "tone": tone,
            "is_neutral": neutral,
            "legacy_family": legacy,
            "aliases": ";".join(aliases.get(key, [])),
        })
    return rows, aliases

ref_rows, aliases = color_rows()
ref_df = pd.DataFrame(ref_rows)
ref_df
```

## 5. Muat dataset dan audit baseColour

Jika `styles.csv` tidak ditemukan, unduh dataset small terlebih dahulu atau unggah file ke Colab.


```python
if not STYLES_CSV.exists():
    raise FileNotFoundError(
        f"{STYLES_CSV} tidak ditemukan. Unduh dataset atau ubah STYLES_CSV."
    )

styles = pd.read_csv(STYLES_CSV, on_bad_lines="warn")
print("baris:", len(styles))
print("kolom:", list(styles.columns))

required = {"id", "articleType", "baseColour", "gender", "usage"}
missing = required - set(styles.columns)
if missing:
    raise ValueError(f"Kolom wajib tidak ada: {sorted(missing)}")
```

```python
base_series = styles["baseColour"].fillna("NA").astype(str).str.strip()
known = set(BASE_TO_FAMILY)
unknown = sorted(set(base_series) - known)
print("baseColour unik:", base_series.nunique())
print("tidak dikenal:", unknown)

if unknown:
    print("Tambahkan ke BASE_TO_FAMILY sebelum ekspor:")
    for value in unknown:
        print(" -", repr(value))
```

## 6. Buat tabel pemetaan dan pratinjau katalog

`base_colour_to_family` menjelaskan setiap label katalog. `styles_color_preview` adalah pratinjau, bukan pengganti katalog final.


```python
base_map = pd.DataFrame([
    {
        "base_colour": base,
        "color_family": family,
        "notes": BASE_NOTES.get(base, ""),
    }
    for base, family in BASE_TO_FAMILY.items()
]).sort_values("base_colour")

styles_color = styles[["id", "articleType", "baseColour", "gender", "usage"]].copy()
styles_color["color_family"] = base_series.map(BASE_TO_FAMILY).fillna("gray")
styles_color["mapping_note"] = base_series.map(BASE_NOTES).fillna("")

base_map.head()
```

## 7. Validasi

Validasi memastikan 16 keluarga unik, semua baseColour terpetakan, dan tidak ada keluarga yang kosong tanpa alasan.


```python
errors = []
if len(ref_df) != 16:
    errors.append("jumlah keluarga bukan 16")
if ref_df["color_family"].duplicated().any():
    errors.append("ada keluarga duplikat")
if ref_df.isna().any().any():
    errors.append("ada nilai kosong di ref_color")
if unknown:
    errors.append(f"baseColour tidak dikenal: {unknown}")

family_counts = styles_color.groupby("color_family").size().sort_values(ascending=False)
zero = [f for f in ref_df["color_family"] if f not in family_counts.index]
if zero:
    print("Keluarga tanpa baseColour:", zero)

print("ERROR:", errors if errors else "tidak ada")
print(family_counts.to_frame("jumlah"))
```

## 8. Ekspor

File keluaran:
- `ml/reports/color/ref_color_16.csv`
- `ml/reports/color/base_colour_to_family.csv`
- `ml/reports/color/styles_color_preview.csv`
- `ml/reports/color/color_summary.json`
- `ml/reports/color/color_reference.db`


```python
ref_path = OUT_DIR / "ref_color_16.csv"
map_path = OUT_DIR / "base_colour_to_family.csv"
preview_path = OUT_DIR / "styles_color_preview.csv"
summary_path = OUT_DIR / "color_summary.json"
db_path = OUT_DIR / "color_reference.db"

ref_df.to_csv(ref_path, index=False)
base_map.to_csv(map_path, index=False)
styles_color.to_csv(preview_path, index=False)

summary = {
    "n_families": int(len(ref_df)),
    "n_base_colours": int(len(base_map)),
    "n_styles_rows": int(len(styles)),
    "unknown_base_colours": unknown,
    "family_counts": {str(k): int(v) for k, v in family_counts.items()},
    "errors": errors,
}
summary_path.write_text(json.dumps(summary, indent=2, ensure_ascii=False), encoding="utf-8")

with sqlite3.connect(db_path) as conn:
    ref_df.to_sql("ref_color", conn, if_exists="replace", index=False)
    base_map.to_sql("base_colour_to_family", conn, if_exists="replace", index=False)

print("ditulis:")
for p in (ref_path, map_path, preview_path, summary_path, db_path):
    print(" -", p.relative_to(ROOT))
print(json.dumps(summary, indent=2, ensure_ascii=False))
```
